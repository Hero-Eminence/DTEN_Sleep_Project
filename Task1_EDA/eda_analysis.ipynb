```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('sleep.csv')
df.head()
```

```python
# How many rows and columns?
print("Shape:", df.shape)

# Column names and data types
print("\nColumns:\n", df.dtypes)

# Any missing values?
print("\nMissing values:\n", df.isnull().sum())
```

```python
# Summary statistics for numeric columns
df.describe()
```

```python
# How many users in each sleep debt category?
print(df['sleep_debt_category'].value_counts())

# Visualize it
sns.set_style("whitegrid")
plt.figure(figsize=(8,5))
sns.countplot(x='sleep_debt_category', data=df, palette='viridis')
plt.title('Distribution of Sleep Debt Categories')
plt.xticks(rotation=15)
plt.savefig('charts/sleep_debt_distribution.png')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='sleep_debt_category', y='total_sleep_hours', data=df, palette='coolwarm')
plt.title('Total Sleep Hours by Sleep Debt Category')
plt.xticks(rotation=15)
plt.savefig('charts/sleep_hours_by_category.png')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='sleep_debt_category', y='bedtime_phone_minutes', data=df, palette='magma')
plt.title('Bedtime Phone Usage by Sleep Debt Category')
plt.xticks(rotation=15)
plt.savefig('charts/phone_usage_by_category.png')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(y='primary_bedtime_app', data=df, order=df['primary_bedtime_app'].value_counts().index, palette='viridis')
plt.title('Most Common Bedtime Apps')
plt.savefig('charts/bedtime_apps.png')
plt.show()
```

```python
plt.figure(figsize=(14,10))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numeric Features')
plt.savefig('charts/correlation_matrix.png')
plt.show()
```

```python
# Encode the target for use in Task 2
df['sleep_debt_encoded'] = df['sleep_debt_category'].map({
    'Optimal Recovery': 0,
    'Mild Deficit': 1,
    'Moderate Debt': 2,
    'Severe Sleep Debt': 3
})

# Save cleaned file
df.to_csv('sleep_cleaned.csv', index=False)
print("Saved sleep_cleaned.csv")
```
